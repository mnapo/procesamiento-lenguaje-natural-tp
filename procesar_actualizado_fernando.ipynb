```python
import json

def analizar_dispersion_longitud(ruta_corpus="datos/corpus.jsonl"):
    """
    Calcula la cantidad de tokens de cada documento
    utilizando len(texto.split()).
    """

    cantidades = []

    with open(ruta_corpus, "r", encoding="utf-8") as archivo:

        for linea in archivo:

            documento = json.loads(linea)

            cantidad_tokens = len(documento["texto"].split())

            cantidades.append(cantidad_tokens)

            print(
                f"{documento['id']} | "
                f"{documento['titulo']} | "
                f"{cantidad_tokens} tokens"
            )

    print("\n--- RESUMEN ---")
    print(f"Cantidad de documentos: {len(cantidades)}")
    print(f"Mínimo: {min(cantidades)} tokens")
    print(f"Máximo: {max(cantidades)} tokens")
    print(f"Diferencia: {max(cantidades) - min(cantidades)} tokens")
    print(f"Promedio: {sum(cantidades) / len(cantidades):.2f} tokens")
analizar_dispersion_longitud()    
```

```python
import json
import re

def revisar_artefactos(ruta_corpus="datos/corpus.jsonl"):
    
    with open(ruta_corpus, "r", encoding="utf-8") as archivo:

        for linea in archivo:
            documento = json.loads(linea)

            texto = documento["texto"]

            html = re.findall(r"<[^>]+>", texto)
            javascript = re.findall(
                r"\b(function|var|let|const|script)\b",
                texto,
                flags=re.IGNORECASE
            )

            espacios_duros = texto.count("\xa0")

            print(
                f"{documento['id']} | "
                f"{documento['titulo']}"
            )

            print(f"Espacios duros: {espacios_duros}")
            print(f"Restos HTML: {len(html)}")
            print(f"Posibles restos JS: {len(javascript)}")
            print("-" * 50)
revisar_artefactos()            
```

```python
import json
import spacy
import pandas as pd

# Cargar modelo de español
nlp = spacy.load("es_core_news_sm")

# Ruta del corpus
ruta_corpus = "datos/corpus.jsonl"

resultados = []

with open(ruta_corpus, "r", encoding="utf-8") as archivo:

    for linea in archivo:

        documento = json.loads(linea)

        texto = documento["texto"]

        # Procesamiento lingüístico con spaCy
        doc = nlp(texto)

        # Solo tokens que sean palabras:
        # excluimos puntuación y espacios
        palabras = [
            token
            for token in doc
            if not token.is_punct and not token.is_space
        ]

        total_palabras = len(palabras)

        # Conteo POS
        cantidad_adjetivos = sum(
            1 for token in palabras
            if token.pos_ == "ADJ"
        )

        cantidad_adverbios = sum(
            1 for token in palabras
            if token.pos_ == "ADV"
        )

        # Frecuencias relativas
        frecuencia_adj = (
            cantidad_adjetivos / total_palabras
            if total_palabras > 0 else 0
        )

        frecuencia_adv = (
            cantidad_adverbios / total_palabras
            if total_palabras > 0 else 0
        )

        resultados.append({
            "id": documento["id"],
            "titulo": documento["titulo"],
            "total_palabras": total_palabras,
            "adjetivos": cantidad_adjetivos,
            "adverbios": cantidad_adverbios,
            "fr_adj": frecuencia_adj,
            "fr_adv": frecuencia_adv,
            "adj_mayor_adv": frecuencia_adj > frecuencia_adv
        })


# Convertir resultados a DataFrame
df_resultados = pd.DataFrame(resultados)

# Mostrar porcentajes con más claridad
df_resultados["fr_adj_pct"] = (
    df_resultados["fr_adj"] * 100
).round(2)

df_resultados["fr_adv_pct"] = (
    df_resultados["fr_adv"] * 100
).round(2)

# Columnas finales para visualizar
columnas_mostrar = [
    "id",
    "titulo",
    "total_palabras",
    "adjetivos",
    "adverbios",
    "fr_adj_pct",
    "fr_adv_pct",
    "adj_mayor_adv"
]

display(df_resultados[columnas_mostrar])
```

Auditoría TTR
Calcular TTR por texto y cruzarlo con longitud para verificar la falacia de que los textos cortos parecen más ricos.
resultado en Tabla/gráfico o análisis TTR vs longitud.

```python
import json
import pandas as pd

# 1. Cargamos directamente el corpus para asegurarnos de tener los textos a mano
datos_corpus = []
with open('datos/corpus.jsonl', 'r', encoding='utf-8') as f:
    for linea in f:
        datos_corpus.append(json.loads(linea))

df_corpus_ttr = pd.DataFrame(datos_corpus)

# 2. Listas para la auditoría de TTR
resultados_ttr = []

# 3. Recorremos cada documento
for i, row in df_corpus_ttr.iterrows():
    texto = row['texto']
    titulo = row['titulo']
    
    # Tokenización estándar por espacios (sin filtrar stopwords para el TTR clásico)
    tokens_ttr = texto.split()
    total_tokens_ttr = len(tokens_ttr)
    
    # Types únicos (pasados a minúsculas)
    types_unicos = set(token.lower() for token in tokens_ttr)
    total_types = len(types_unicos)
    
    # Cálculo del Type-Token Ratio
    ttr_valor = total_types / total_tokens_ttr if total_tokens_ttr > 0 else 0
    
    resultados_ttr.append({
        'Título': titulo,
        'Tokens': total_tokens_ttr,
        'Types_Unicos': total_types,
        'TTR': ttr_valor
    })

# DataFrame exclusivo para el análisis de TTR ordenado por cantidad de tokens
df_auditoria_ttr = pd.DataFrame(resultados_ttr)

print("--- Auditoría TTR vs Longitud ---")
display(df_auditoria_ttr.sort_values(by='Tokens', ascending=True))
```

Observaciones:
Tokens: Son todas las palabras que componen el texto, una por una, contando absolutamente todas las apariciones (incluso si se repiten). Por ejemplo, si la palabra "el" aparece 50 veces en una novela, cuenta como 50 tokens distintos dentro del total de palabras del libro.

Types_Unicos (o Types): Son las palabras diferentes o únicas que forman el vocabulario de ese texto, ignorando por completo cuántas veces se repiten. Siguiendo el ejemplo anterior, no importa si "el" apareció 50 veces; como type solo cuenta como 1 palabra única en el inventario del vocabulario.

TTR (Type-Token Ratio): Es la división matemática entre ambos (Types / Tokens). Esta métrica mide la variedad o "riqueza léxica" del texto.
Entonces:
    Si un texto tiene un TTR cercano a 1 (o 100%), significa que casi todas sus palabras son diferentes y casi no hay repeticiones (típico de textos muy cortos).
    Si tiene un TTR bajo (cercano a 0), significa que el texto recicla mucho las mismas palabras (artículos, verbos, pronombres) a lo largo de miles de tokens, lo cual es normal en novelas más extensas.

Conclusión:
Al analizar los resultados de la auditoría de TTR en relación con la longitud de los textos, se verifica que los textos ordenados por cantidad de tokens, de menor a mayor, muestran una relación inversamente proporcional con su TTR.
Entonces:
Las obras con menor cantidad de tokens presentan los valores de TTR más elevados (acercándose a proporciones de vocabulario único más altas), mientras que los textos más extensos registran un TTR considerablemente menor. Esto demuestra empíricamente que el TTR sin normalizar penaliza a los textos largos —debido a la constante repetición natural de palabras funcionales y estructurales como artículos, preposiciones y verbos auxiliares—, generando el sesgo erróneo de que los textos breves poseen una mayor riqueza o diversidad léxica por el simple hecho de ser cortos.

Auditoría TF-IDF
Explicar y comprobar en el corpus el efecto de smooth_idf=True y la norma L2; revisar por qué términos presentes en todos los documentos no necesariamente dan 0.
Comprobación sklearn vs cálculo/explicación de clase.

```python
import json
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
import numpy as np

# 1. Cargamos directamente el corpus del archivo para asegurar la independencia de la celda
datos_corpus = []
with open('datos/corpus.jsonl', 'r', encoding='utf-8') as f:
    for linea in f:
        datos_corpus.append(json.loads(linea))

df_corpus = pd.DataFrame(datos_corpus)

textos = df_corpus['texto'].tolist()
titulos = df_corpus['titulo'].tolist()

# 2. Vectorización TF-IDF con los parámetros exigidos por la auditoría (smooth_idf=True y norm='l2')
vectorizer = TfidfVectorizer(smooth_idf=True, norm='l2')
matriz_tfidf = vectorizer.fit_transform(textos)

# 3. Comprobación de smooth_idf: analizamos los valores IDF calculados por sklearn
df_idf = pd.DataFrame({
    'Palabra': vectorizer.get_feature_names_out(),
    'IDF': vectorizer.idf_
})

print("--- Auditoría TF-IDF: Comprobación de smooth_idf (ningún IDF da 0) ---")
display(df_idf.sort_values(by='IDF', ascending=True).head(10))

# 4. Comprobación de la Norma L2 por documento (cada vector de fila debe normalizarse a 1.0)
normas_filas = np.linalg.norm(matriz_tfidf.toarray(), axis=1)
df_normas = pd.DataFrame({'Título': titulos, 'Norma_L2': normas_filas})

print("\n--- Auditoría TF-IDF: Verificación de Normalización L2 (vectores unitarios) ---")
display(df_normas)
```

 TF-IDF (Term Frequency - Inverse Document Frequency) es una herramienta matemática que sirve para descubrir qué palabras son verdaderamente importantes en un texto dentro de un conjunto de documentos, filtrando las que son comunes. 
 Se compone de dos partes:  
 TF (Frecuencia del término): Mide cuántas veces aparece una palabra dentro de ese mismo texto (si se repite mucho, probablemente sea central ahí).
 IDF (Frecuencia inversa de documento): Mide qué tan rara o común es esa palabra en todo el corpus. Si una palabra aparece en todos lados (como "el" o "de"), su valor baja; si aparece en pocos textos, es una palabra "rara" y su valor sube.  
 ¿Qué demuestran los parámetros por separado y juntos?
 1. Por separado: El efecto de smooth_idf=True (la primera tabla, en 1.0)
 Qué hace: En la teoría matemática tradicional, si una palabra aparece en el 100% de los documentos, su cálculo de IDF debería dar cero (lo que borraría por completo su existencia en el análisis).  
 Qué demuestra: Como scikit-learn aplica un sistema de suavizado interno por diseño, le suma un ajuste a la fórmula para que las palabras ultra comunes no mueran valiendo cero. 
 El resultado 1.0 en la tabla demuestra justamente eso: la librería establece un "piso mínimo" de importancia para que los términos transversales no desaparezcan. 

  Por separado: El efecto de la Norma L2 (Tu segunda tabla en 1.0)
    Qué hace: Agarra el vector numérico de cada documento y lo escala por su propia longitud para que el total de cada fila dé exactamente 1.0.
    Qué demuestra: Elimina el sesgo de longitud. Prueba que un libro de 500 páginas no va a tener más peso matemático que un cuento de 2 páginas solo por tener más palabras escritas; los pone a todos a competir en un mismo "escenario nivelado"

Combinados (El resultado final)
Juntos transforman un texto desordenado en un perfil matemático limpio y justo: smooth_idf evita que las palabras comunes se anulen por completo, y la norma L2 evita que los textos largos dominen a los cortos
