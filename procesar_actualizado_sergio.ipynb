```python
from scraper_actualizado import Scraper

URLS = [
    {
        "url": "https://es.wikisource.org/wiki/Historia_de_un_amor_turbio",
        "multipagina": False
    },
    {
        "url": "https://es.wikisource.org/wiki/A_flor_de_piel",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/A_fuego_lento",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/A_los_pies_de_Venus",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/A_prueba",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/A_todo_honor",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Adem%C3%A1s_del_frac",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Aita_Tettauen",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Al_amor_de_los_tizones",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Al_primer_vuelo",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Amadeo_I",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Amalia",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Amistad_funesta",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Ana_Karenina",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/%C3%81ngel_Guerra_(Gald%C3%B3s)",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Arroz_y_tartana",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/As%C3%AD_paga_el_diablo",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Aura_o_las_violetas",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Aurora_roja",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Baile_y_cochino",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Bail%C3%A9n_(P%C3%A9rez_Gald%C3%B3s)",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Blasones_y_talegas",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Bodas_reales",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Brenda",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Buenos_Aires_en_el_a%C3%B1o_2080",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Bug-Jargal",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/C%C3%A1diz",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Camila_o_la_virtud_triunfante",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/C%C3%A1novas",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Caramur%C3%BA",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Carlos_VI_en_la_R%C3%A1pita",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Ca%C3%B1as_y_barro",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Cel%C3%ADn",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Cinco_semanas_en_globo",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Clemencia_(Altamirano)",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Clemencia_(Caballero)",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Crimen_y_castigo_(tr._an%C3%B3nima)",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Cr%C3%B3nica_del_reinado_de_Carlos_IX",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Cuesti%C3%B3n_de_ambiente",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Cumand%C3%A1",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/De_Cartago_a_Sagunto",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/De_O%C3%B1ate_a_la_Granja",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/De_tal_palo,_tal_astilla",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Del_fr%C3%ADo_al_fuego",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Don_Gonzalo_Gonz%C3%A1lez_de_la_Gonzalera",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Don_Segundo_Sombra",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Do%C3%B1a_Luz",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Do%C3%B1a_Milagros",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Do%C3%B1a_Perfecta",
        "multipagina": True
    },
    {
        "url": "https://es.wikisource.org/wiki/Dulce_due%C3%B1o_(Pardo_Baz%C3%A1n)",
        "multipagina": True
    }
]

NOMBRE_DE_CARPETA = "datos"

CABECERAS = {
    "User-Agent": "LabPLN-2026/1.0 (contacto: tu-mail@ejemplo.com)"
}

STOPWORDS = r"c:\Ciencia De Datos\NLP\pair_programming_ia\stopwords_es.txt"


def cargar_stopwords(ruta_archivo):
    """Carga el archivo de stopwords y devuelve un conjunto de palabras."""
    with open(ruta_archivo, "r", encoding="utf-8") as f:
        return {
            linea.strip().lower()
            for linea in f
            if linea.strip()
        }


def inicializar_scraper(url, nombre_archivo):
    """
    Inicializa el scraper, verifica robots.txt,
    obtiene la página principal y guarda su HTML.
    """
    scraper = Scraper(
        url,
        headers=CABECERAS,
        nombre_archivo=nombre_archivo,
        carpeta_destino=NOMBRE_DE_CARPETA
    )

    permitido = scraper.permite_scrapeo()

    if permitido:
        print(f"Scrapeo permitido por robots.txt: {permitido}")
        scraper.actualizo_csv(url, "robots", "OK")

        respuesta = scraper.obtener_pagina()

        if respuesta:
            print("Scrapeo exitoso")
            scraper.guardar_pagina()
            scraper.actualizo_csv(url, "extraccion", "OK")
            return scraper

        print("No se pudo obtener la página")
        scraper.actualizo_csv(url, "extraccion", "Error")
        return False

    print("Scrapeo no permitido")
    scraper.actualizo_csv(url, "robots", "No permitido")

    print(
        "Forzar scrapeo sin autorización de robots.txt "
        "(no recomendado) s/n:"
    )
    respuesta = input().lower()

    if respuesta == "s":
        print("Forzando scrapeo sin autorización de robots.txt")

        respuesta = scraper.obtener_pagina()

        if respuesta:
            print("Scrapeo exitoso")
            scraper.guardar_pagina()
            scraper.actualizo_csv(url, "extraccion", "OK")
            return scraper

    print("Scrapeo cancelado")
    scraper.actualizo_csv(url, "extraccion", "Cancelado")

    return False


def contar_palabra(scraper, palabra):
    """Cuenta cuántas veces aparece una palabra en la página principal."""
    cantidad = scraper.contar_palabra(palabra)

    print(
        f"La palabra '{palabra}' aparece "
        f"{cantidad} veces en la página."
    )


def main():
    """Recorre todas las obras definidas en URLS y arma el corpus."""

    stopwords = cargar_stopwords(STOPWORDS)

    for i, item in enumerate(URLS, start=1):

        url = item["url"]
        multipagina = item["multipagina"]

        print("\n" + "=" * 60)
        print(f"Procesando URL {i}:")
        print(url)
        print(f"Multipágina: {multipagina}")
        print("=" * 60)

        nombre_archivo = f"pagina_{i}.html"

        scraper = inicializar_scraper(
            url,
            nombre_archivo
        )

        if not scraper:
            print(
                "Esta URL no pudo ser procesada. "
                "Se continúa con la siguiente."
            )
            continue

        scraper.actualizo_csv(
            url,
            "multipagina",
            "Sí" if multipagina else "No"
        )

        # Los metadatos se extraen siempre de la página principal.
        datos_corpus = scraper.obtener_datos_corpus()
        
       
        scraper.actualizo_csv(
            url,
            "ruta",
            str(scraper.ruta_archivo)
        )

        scraper.actualizo_csv(
            url,
            "Titulo de la página",
            datos_corpus.get("title", "")
        )

        scraper.actualizo_csv(
            url,
            "Autoría de la página",
            datos_corpus.get("author", "")
        )

        scraper.actualizo_csv(
            url,
            "Fecha de publicación de la página",
            datos_corpus.get("date", "")
        )
        
        fecha_publicacion_obra = scraper.obtener_fecha_publicacion_obra()
        
        scraper.actualizo_csv(
            url,
            "Año de publicación de la Obra",
            fecha_publicacion_obra
        )
        
        print(f"Fecha de publicación obtenida: {fecha_publicacion_obra}")
        
        scraper.actualizo_csv(
            url,
            "Medio",
            datos_corpus.get("hostname", "")
        )

        # --------------------------------------------------------
        # ACÁ SE DECIDE EXPLÍCITAMENTE QUÉ TIPO DE SCRAPING USAR.
        # --------------------------------------------------------
        if multipagina:
            texto_corpus = scraper.obtener_texto_multipagina()
            
        else:
            texto_corpus = datos_corpus.get("text", "")

            
        if not texto_corpus:
            print("No se pudo extraer texto del corpus.")
            scraper.actualizo_csv(
                url,
                "extraccion",
                "Sin texto"
            )
            continue

        datos_normalizados = scraper.normalizar(
            texto_corpus
        )

        scraper.actualizo_csv(
            url,
            "normalizacion",
            "OK"
        )
        resultado = scraper.tokenizar_documentos([datos_normalizados])
        print("Resultado de la tokenización:", resultado)
        scraper.actualizo_csv(
            url,
            "Cant-Tokens",
            resultado
        )
        
        scraper.actualizo_csv(
            url,
            "Licencia",
            datos_corpus.get("license", "")
        )
        
        datos_sin_stopwords = scraper.palabras_significativas(
            texto_corpus,
            stopwords
        )

        print(
            "Datos del corpus sin stopwords: "
            f"{datos_sin_stopwords[:100]}..."
        )

        scraper.guardar_datos_corpus(
            datos_normalizados,
            tipo_dato="texto",
            tipo_archivo="normalizado"
        )
        
        scraper.actualizo_csv(
            url,
            "guardado",
            "OK"
        )    

if __name__ == "__main__":
    main()
```

```python
import json

def analizar_dispersion_longitud(ruta_corpus="datos/corpus.jsonl"):
    """
    Calcula la cantidad de tokens de cada documento
    utilizando len(texto.split()).
    """

    cantidades = []

    with open(ruta_corpus, "r", encoding="utf-8") as archivo:

        for linea in archivo:

            documento = json.loads(linea)

            cantidad_tokens = len(documento["texto"].split())

            cantidades.append(cantidad_tokens)

            print(
                f"{documento['id']} | "
                f"{documento['titulo']} | "
                f"{cantidad_tokens} tokens"
            )

    print("\n--- RESUMEN ---")
    print(f"Cantidad de documentos: {len(cantidades)}")
    print(f"Mínimo: {min(cantidades)} tokens")
    print(f"Máximo: {max(cantidades)} tokens")
    print(f"Diferencia: {max(cantidades) - min(cantidades)} tokens")
    print(f"Promedio: {sum(cantidades) / len(cantidades):.2f} tokens")
analizar_dispersion_longitud()    
```

```python
import json
import re

def revisar_artefactos(ruta_corpus="datos/corpus.jsonl"):
    
    with open(ruta_corpus, "r", encoding="utf-8") as archivo:

        for linea in archivo:
            documento = json.loads(linea)

            texto = documento["texto"]

            html = re.findall(r"<[^>]+>", texto)
            javascript = re.findall(
                r"\b(function|var|let|const|script)\b",
                texto,
                flags=re.IGNORECASE
            )

            espacios_duros = texto.count("\xa0")

            print(
                f"{documento['id']} | "
                f"{documento['titulo']}"
            )

            print(f"Espacios duros: {espacios_duros}")
            print(f"Restos HTML: {len(html)}")
            print(f"Posibles restos JS: {len(javascript)}")
            print("-" * 50)
revisar_artefactos()            
```

```python
# pandas: para trabajar el corpus como tabla.
import pandas as pd

# Counter: para contar cosas. Devuelve un diccionario palabra -> veces.
from collections import Counter

corpus = pd.read_json("datos/corpus.jsonl", lines=True)

print("textos:", f"{len(corpus):,}")
```

```python
# string: trae la lista de signos de puntuación.
import string

signos_del_espanol = "¡¿—–…“”‘’«»"
todos_los_signos = string.punctuation + signos_del_espanol

# re.escape prepara los signos para que la expresión regular los tome literalmente.
patron_de_signos = "[" + re.escape(todos_los_signos) + "]"

def limpiar(texto):
    # Pasa a minúsculas, saca la puntuación y unifica los espacios.
    # NO saca números: en este corpus los años son datos.
    en_minusculas = texto.lower()
    sin_signos = re.sub(patron_de_signos, " ", en_minusculas)
    sin_espacios_de_mas = re.sub(r"\s+", " ", sin_signos)
    return sin_espacios_de_mas.strip()
```

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizador = TfidfVectorizer(stop_words=palabras_vacias, min_df=2)
pesos = vectorizador.fit_transform(documentos_por_id)

vocabulario = vectorizador.get_feature_names_out()
tabla_pesos = pesos.toarray()

# Para cada documento, las seis palabras con más peso.
filas = []
for posicion in range(len(docus)):
    pesos_del_documento = tabla_pesos[posicion]
    orden = np.argsort(pesos_del_documento)
    mejores = []
    for indice in orden[-6:]:
        mejores.append(vocabulario[indice])
    mejores.reverse()
    fila = {"documento": docus[posicion], "palabras_con_mas_peso": ", ".join(mejores)}
    filas.append(fila)

pd.DataFrame(filas)
```

```python
import json
import spacy
import pandas as pd

# Cargar modelo de español
nlp = spacy.load("es_core_news_sm")

# Ruta del corpus
ruta_corpus = "datos/corpus.jsonl"

resultados = []

with open(ruta_corpus, "r", encoding="utf-8") as archivo:

    for linea in archivo:

        documento = json.loads(linea)

        texto = documento["texto"]

        # Procesamiento lingüístico con spaCy
        doc = nlp(texto)

        # Solo tokens que sean palabras:
        # excluimos puntuación y espacios
        palabras = [
            token
            for token in doc
            if not token.is_punct and not token.is_space
        ]

        total_palabras = len(palabras)

        # Conteo POS
        cantidad_adjetivos = sum(
            1 for token in palabras
            if token.pos_ == "ADJ"
        )

        cantidad_adverbios = sum(
            1 for token in palabras
            if token.pos_ == "ADV"
        )

        # Frecuencias relativas
        frecuencia_adj = (
            cantidad_adjetivos / total_palabras
            if total_palabras > 0 else 0
        )

        frecuencia_adv = (
            cantidad_adverbios / total_palabras
            if total_palabras > 0 else 0
        )

        resultados.append({
            "id": documento["id"],
            "titulo": documento["titulo"],
            "total_palabras": total_palabras,
            "adjetivos": cantidad_adjetivos,
            "adverbios": cantidad_adverbios,
            "fr_adj": frecuencia_adj,
            "fr_adv": frecuencia_adv,
            "adj_mayor_adv": frecuencia_adj > frecuencia_adv
        })


# Convertir resultados a DataFrame
df_resultados = pd.DataFrame(resultados)

# Mostrar porcentajes con más claridad
df_resultados["fr_adj_pct"] = (
    df_resultados["fr_adj"] * 100
).round(2)

df_resultados["fr_adv_pct"] = (
    df_resultados["fr_adv"] * 100
).round(2)

# Columnas finales para visualizar
columnas_mostrar = [
    "id",
    "titulo",
    "total_palabras",
    "adjetivos",
    "adverbios",
    "fr_adj_pct",
    "fr_adv_pct",
    "adj_mayor_adv"
]

display(df_resultados[columnas_mostrar])
```

```python
documentos_cumplen = df_resultados[
    "adj_mayor_adv"
].sum()

total_documentos = len(df_resultados)

print(
    f"Documentos donde ADJ > ADV: "
    f"{documentos_cumplen} de {total_documentos}"
)

print(
    f"Documentos donde ADJ <= ADV: "
    f"{total_documentos - documentos_cumplen}"
)

promedio_adj = (
    df_resultados["fr_adj"].mean() * 100
)

promedio_adv = (
    df_resultados["fr_adv"].mean() * 100
)

print(
    f"Frecuencia relativa media de adjetivos: "
    f"{promedio_adj:.2f}%"
)

print(
    f"Frecuencia relativa media de adverbios: "
    f"{promedio_adv:.2f}%"
)
```
